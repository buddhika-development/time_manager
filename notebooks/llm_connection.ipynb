```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_mistralai import  ChatMistralAI
from abc import ABC, abstractmethod
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
class LLMConnection(ABC):
    def __init__(self):
        self.llm = None

    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def invoke(self, message):
        pass
```

```python
class GEMINIConnection(LLMConnection):
    def __init__(self):
        super().__init__()
        self.connect()
    
    def  connect(self):
        self.llm = ChatGoogleGenerativeAI(
            model= "models/gemini-2.5-flash",
            google_api_key = os.getenv("GOOGLE_API_KEY"),
        )

    def invoke(self, message):
        return self.llm.invoke(message)
```

```python
class MISTRALConnection(LLMConnection):
    def __init__(self):
        super().__init__()
        self.connect()

    def connect(self):
        self.llm = ChatMistralAI(
            model = "mistral-large-2512",
        )

    def invoke(self, message):
        return self.llm.invoke(message)
```

```python
class GetLLM():

    _connections = {
        "gemini": GEMINIConnection,
        "mistral": MISTRALConnection
    }

    @classmethod
    def create(cls, provider:str, **kwargs) -> LLMConnection:
        if provider.lower() not in cls._connections:
            raise ValueError(f"Invalid provider: {provider}")
        return cls._connections[provider.lower()](**kwargs)
```

```python
llm = GetLLM.create("gemini")
llm.invoke("Hello").content
```
